### Question 8
Using the diamonds dataset and the plotting code of your choice, plot the following from the diamonds dataset.

1. Carat vs Price in a scatter plot.
2. A histogram of prices.
3. A boxplot of cut vs price.

The code can be loaded from seaborn in the cell below.

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

diamonds = sns.load_dataset('diamonds')
diamonds

#Code to create a Scatter Plot of Diamond Carat vs Price
diamonds.plot.scatter(x='carat', y='price', color = 'skyblue')
plt.title('Carat vs Price of Diamonds')
plt.xlabel('Carat')
plt.ylabel('Price')
plt.show()
````

````python
#Code to create a Histogram of Diamond Prices
diamonds['price'].plot.hist(bins=10, color='midnightblue', edgecolor = 'lavender')
plt.title('Distribution of Diamond Price')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.show()
````

````python
#Code to create a Boxplot of Diamond Price distribution between Cuts
#Removes heading 'Boxplot grouped by cut'
diamonds.boxplot(column='price', by='cut', grid = False)
plt.title('Distribution of Diamond Price by Cut')
plt.suptitle('')  
plt.xlabel('Cut')
plt.ylabel('Price')
plt.show()
````
